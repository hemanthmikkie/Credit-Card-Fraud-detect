# 04. Feature Engineering & Leakage Prevention

## Objectives:
1. Extract temporal cyclical features (`sin_hour`, `cos_hour`).
2. Temper right-skewed amounts via `log_amount = log1p(Amount)`.
3. Calculate relative amount ratio relative to baseline training median.
4. Encapsulate transformations in a Scikit-Learn compatible transformer to guarantee zero data leakage.


```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
import numpy as np
from src.config import CLEANED_DATA_PATH
from src.feature_engineering import FraudFeatureEngineer

df = pd.read_csv(CLEANED_DATA_PATH)
fe = FraudFeatureEngineer()
fe.fit(df)
df_fe = fe.transform(df)

print('Engineered features created:', [c for c in df_fe.columns if c not in df.columns])
df_fe[['Time', 'hour_of_day', 'sin_hour', 'cos_hour', 'Amount', 'log_amount', 'amount_ratio']].head()
```

### Mathematical Verification of Cyclical Continuity


```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 6))
plt.scatter(df_fe['sin_hour'].iloc[:600], df_fe['cos_hour'].iloc[:600], color='#2563EB', alpha=0.5)
plt.title('Cyclical 24-Hour Projection (sin_hour vs cos_hour)', fontweight='bold')
plt.xlabel('sin(2 * pi * hour / 24)')
plt.ylabel('cos(2 * pi * hour / 24)')
plt.grid(True)
plt.show()
```
