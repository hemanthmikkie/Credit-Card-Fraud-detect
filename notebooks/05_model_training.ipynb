```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
from src.config import CLEANED_DATA_PATH
from src.preprocessing import prepare_data_pipeline, apply_smote
from src.train_model import train_and_compare_models, save_model_artifacts

df = pd.read_csv(CLEANED_DATA_PATH)
X_train, X_test, y_train, y_test, artifact = prepare_data_pipeline(df)
print(f'Train split: {len(X_train):,} rows | Test split: {len(X_test):,} rows')
```

### Applying SMOTE Strictly to Training Split

**CRITICAL RULE**: SMOTE must NEVER be applied to test data. Doing so synthetically inflates evaluation scores.


```python
X_train_res, y_train_res = apply_smote(X_train, y_train, sampling_strategy=0.10)
print(f'Resampled Train: {len(X_train_res):,} samples (Fraud Ratio: {y_train_res.mean()*100:.2f}%)')
```
